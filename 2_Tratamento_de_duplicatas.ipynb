```python
import pandas as pd
import os
```

```python
pasta_dados_processados = "./processed"
arquivos_na_pasta = os.listdir(pasta_dados_processados)
```

```python
for arquivo in arquivos_na_pasta:
    if arquivo.endswith('.parquet'):
        caminho_parquet = os.path.join(pasta_dados_processados, arquivo)
        df = pd.read_parquet(caminho_parquet)
        print(f"Processando {arquivo}")
        total_duplicadas = df.duplicated().sum()
        print(f"Total de duplicatas: {total_duplicadas}")
    else:
        print("Não é um arquivo .parquet.") 
```
