```python
import pandas as pd
import os
```

```python
# Define pastas de dados brutos e processados
pasta_dados_brutos = "./raw"
pasta_dados_processados = "./processed"
```

```python
# Garante que a pasta destino existe
os.makedirs(pasta_dados_processados, exist_ok=True)
```

```python
# Verifica arquivos dentro da pasta dados brutos e cria lista com nomes
arquivos_na_pasta = os.listdir(pasta_dados_brutos)
print(arquivos_na_pasta)
```

```python
for arquivo in arquivos_na_pasta:
    if arquivo.endswith('.csv'):
        caminho_csv = os.path.join(pasta_dados_brutos, arquivo)
        df = pd.read_csv(caminho_csv)
        arquivo_parquet = arquivo.replace('.csv','.parquet')
        caminho_parquet = os.path.join(pasta_dados_processados, arquivo_parquet)
        df.to_parquet(caminho_parquet, engine='fastparquet', compression='snappy')
        print(f"Salvo com sucesso em: {caminho_parquet}\n")
    else:
        print("Nenhum .csv foi encontrado ") 
```

```python
arquivos_para_renomear = os.listdir(pasta_dados_processados)
lista_nomes_novos = ['clientes', 'geolocalizacao', 'pagamento_tratado', 'itens_pedidos', 'pagamentos_pedidos', 'produtos', 'vendedores', 'geolocalizacao_unicas', 'traducao_categoria']
lista_arquivos_parquet = [arquivo for arquivo in arquivos_para_renomear if arquivo.endswith('.parquet')]
lista_arquivos_parquet
```

```python
for nome_antigo, nome_novo in zip(lista_arquivos_parquet, lista_nomes_novos): 
    caminho_antigo = os.path.join(pasta_dados_processados, nome_antigo)
    nome_novo_com_extensao = nome_novo + '.parquet'

    caminho_novo = os.path.join(pasta_dados_processados, nome_novo_com_extensao)

    os.rename(caminho_antigo, caminho_novo)
    print(f"Renomeado: {nome_antigo} -> {nome_novo_com_extensao}")
```

```python
#Checagem de tamanho das tabelas parquet e csv

for arquivo in arquivos_na_pasta:
    if arquivo.endswith('.csv'):
        caminho_csv = os.path.join(pasta_dados_brutos, arquivo)
        df_shape = pd.read_csv(caminho_csv).shape
        print(df_shape)
```
